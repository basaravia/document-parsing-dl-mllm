```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(
    model="ai/qwen3",
    base_url="http://localhost:12434/engines/llama.cpp/v1",
    api_key="dummy",
)

print(llm.invoke([HumanMessage(content="Di hola y tu nombre de modelo.")]).content)
```

```python
queries = [
    [HumanMessage(content="Define Docker Model Runner en 1 línea.")],
    [HumanMessage(content="Dame 3 casos de uso de DMR.")],
    [HumanMessage(content="¿Cómo listar modelos disponibles?")]
]

for out in llm.batch(queries):
    print("-" * 40)
    print(out.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

lm = ChatOpenAI(
    model="ai/qwen3",               # modelo un poco más grande
    temperature=0.2,
    max_tokens=256,                 # controla longitud de salida
    timeout=60,                     # segundos
    max_retries=2,                  # reintentos ante errores transitorios
    api_key="dummy",
    base_url="http://localhost:12434/engines/llama.cpp/v1"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "Eres un experto en {tema}. Responde en español, breve."),
    ("human", "{pregunta}")
])

chain = prompt | llm
print(chain.invoke({"tema": "Docker y MLOps", "pregunta": "¿Qué es docker model y para qué sirve?"}).content)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.callbacks import CallbackManager
from langchain_core.callbacks.streaming_stdout import StreamingStdOutCallbackHandler

llm = ChatOpenAI(
    model="ai/qwen3",               # modelo un poco más grande
    temperature=0.2,
    max_tokens=1000,                 # controla longitud de salida
    timeout=60,                     # segundos
    max_retries=2,                  # reintentos ante errores transitorios
    api_key="dummy",
    base_url="http://localhost:12434/engines/llama.cpp/v1",
    streaming=True,
    callback_manager=CallbackManager([StreamingStdOutCallbackHandler()]),
)

llm.invoke([HumanMessage(content="Resume qué es Docker Model Runner en un ensayo.")])
# Los tokens se imprimen a medida que llegan
```
